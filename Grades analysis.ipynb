Great, now lets do a quick time plot

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame.from_dict(grades, orient="index")

df.columns = [ "Subject", "Number", "Credits", "Grade", "Class AVG", "Class Size", "Year", "Semester", "Art", "Sci", "Type" ]

df.index.name = "Course"
df = df.reset_index()

year_order = { "First": 1, "Second": 2, "Third": 3, "Fourth": 4, "Fifth": 5 }

df["Year_Num"] = df["Year"].map(year_order)

df["Time"] = df["Year_Num"] * 10 + df["Semester"] # convert semester to chronological position

df = df.sort_values("Time")# sort chronologically

df.drop(df.tail(1).index,inplace=True) # drop last n rows

df
```

```python
plt.figure(figsize=(16, 7))

sns.lineplot(
    data=df,
    x="Time",
    y="Grade",
    marker="o",
    label="My Grade"
)

sns.lineplot(
    data=df,
    x="Time",
    y="Class AVG",
    marker="o",
    label="Class Average"
)

plt.title("Academic Performance Over Time")
plt.xlabel("Academic Term")
plt.ylabel("Grade (%)")
plt.ylim(50, 100)
plt.grid(True, alpha=0.3)
plt.legend()
plt.tight_layout()
plt.show()
```

# Now lets make some specific plots:

### Course scatter plot:

```python
plt.figure(figsize=(16, 8))

sns.scatterplot(
    data=df,
    x="Course",
    y="Grade",
    size="Credits",
    sizes=(50, 300),
    alpha=0.75,
    legend="brief"
)

plt.title("Grades Across All Courses")
plt.xlabel("Course")
plt.ylabel("Grade (%)")
plt.ylim(50, 100)
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Major vs Minor scatter plot:

```python
plt.figure(figsize=(16, 8))

sns.scatterplot(
    data=df,
    x="Course",
    y="Grade",
    size="Credits",
    sizes=(50, 300),
    hue="Type",
    palette={"Major": "green", "Minor": "purple", "Elective": "blue"},
    alpha=0.8
)

plt.title("Grades Across All Courses by Degree Classification")
plt.xlabel("Course")
plt.ylabel("Grade (%)")
plt.ylim(50, 100)
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

### Art vs Science scatter plot:

```python
plt.figure(figsize=(16, 8))

sns.scatterplot(
    data=df,
    x="Course",
    y="Grade",
    size="Credits",
    sizes=(50, 300),
    hue="Art",
    palette={True: "red", False: "blue"},
    alpha=0.8
)

plt.title("Grades Across All Courses by Faculty")
plt.xlabel("Course")
plt.ylabel("Grade (%)")
plt.ylim(50, 100)
plt.xticks(rotation=90)
plt.grid(True, alpha=0.3)

# Rename legend
handles, labels = plt.gca().get_legend_handles_labels()
labels = ["Arts" if label == "True" else "Science" if label == "False" else label for label in labels]
plt.legend(handles, labels, title="Faculty / Credits")

plt.tight_layout()
plt.show()
```

### A butterfly plot for specific course grades:

```python
year_order = {"First": 1, "Second": 2, "Third": 3, "Fourth": 4, "Fifth": 5}

df["Year_Num"] = df["Year"].map(year_order)

terms = {
    "Winter Term 1": ([1, 3], "red"),
    "Winter Term 2": ([2, 3], "cyan"),
    "Summer": ([4, 5, 6], "yellow")
}

fig, axes = plt.subplots(1, 15, figsize=(30, 7))
plot_index = 0

for year in sorted(df["Year"].unique(), key=lambda x: year_order[x]):

    year_df = df[df["Year"] == year]

    for term_name, (semesters, color) in terms.items():
        ax = axes[plot_index]
        term_df = year_df[year_df["Semester"].isin(semesters)]

        if not term_df.empty:
            sns.boxplot(data=term_df, y="Grade", color=color, width=0.4, ax=ax)
            sns.stripplot(data=term_df, y="Grade", color="black", size=6, jitter=0.12, ax=ax)

        ax.set_title(f"{year}\n{term_name}")
        ax.set_ylabel("Grade (%)")
        ax.set_ylim(50, 100)
        ax.grid(axis="y", alpha=0.3)

        plot_index += 1

plt.suptitle("Grade Distribution Across All Years and Terms", fontsize=18)
plt.tight_layout()
plt.show()
```
